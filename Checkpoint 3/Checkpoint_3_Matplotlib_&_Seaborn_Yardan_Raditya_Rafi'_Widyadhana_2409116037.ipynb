# **Library**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Data Visualization**

## **Membaca Data**

```python
df = pd.read_csv("/content/student_performance_dataset.csv")

df
```

## **Bar Chart**

```python
# Set ukuran figure
plt.figure(figsize=(8, 5))

# Plot Bar Chart
sns.countplot(x="LearningStrategy", data=df, palette="Blues_r")

# Tambahkan judul dan label
plt.title("Distribusi Strategi Pembelajaran Siswa", fontsize=14)
plt.xlabel("Strategi Pembelajaran", fontsize=12)
plt.ylabel("Jumlah Siswa", fontsize=12)

# Tampilkan plot
plt.show()
```

Di diagram ini terlihat bahwa pola pembelajaran Minimalist memiliki jumlah siswa paling banyak dibandingkan dua pola lainnya. Hal ini menunjukkan bahwa mayoritas siswa memiliki waktu belajar yang relatif rendah dengan dukungan minimal dari bimbingan tambahan atau orang tua.

Sementara itu, pola pembelajaran Intensive berada di posisi kedua, yang berarti masih ada cukup banyak siswa yang menghabiskan waktu belajar lebih lama serta mendapatkan dukungan tambahan.

Di sisi lain, pola Balanced memiliki jumlah siswa paling sedikit, yang menunjukkan bahwa hanya sedikit siswa yang memiliki keseimbangan antara waktu belajar dan kegiatan ekstrakurikuler.

## **Box Plot**

```python
plt.figure(figsize=(8,6))
sns.boxplot(x='LearningStrategy', y='GPA', data=df, palette="pastel")
plt.xlabel("Strategi Pembelajaran")
plt.ylabel("GPA")
plt.title("Distribusi GPA Berdasarkan Strategi Pembelajaran")
plt.show()
```

Box plot di atas menggambarkan bagaimana distribusi nilai GPA (Grade Point Average) berbeda berdasarkan tiga strategi pembelajaran yang digunakan oleh siswa, yaitu Intensive, Minimalist, dan Balanced. Grafik ini membantu memahami bagaimana setiap strategi memengaruhi nilai akademik siswa secara keseluruhan.

Dari grafik ini, dapat dilihat bahwa setiap strategi memiliki penyebaran nilai yang cukup luas, dengan rentang GPA yang berkisar dari 0 hingga 4. Hal ini menunjukkan bahwa ada variasi yang cukup besar dalam performa akademik siswa, terlepas dari strategi pembelajaran yang mereka gunakan.

## **Pie Chart**

```python
# Data jumlah siswa berdasarkan pola pembelajaran
labels = ['Intensive', 'Minimalist', 'Balanced']
values = [700, 1450, 250]
colors = ['#1f77b4', '#aec7e8', '#c7c7c7']

# Membuat pie chart
plt.figure(figsize=(7, 7))
plt.pie(values, labels=labels, autopct='%1.1f%%', colors=colors, startangle=140, shadow=True)
plt.title('Distribusi Strategi Pembelajaran Siswa')
plt.show()
```

Pie chart ini menunjukkan distribusi pola pembelajaran siswa dalam bentuk persentase. Dari diagram ini, terlihat bahwa mayoritas siswa menggunakan pola pembelajaran Minimalist dengan proporsi 60.4%, menunjukkan bahwa sebagian besar siswa memiliki waktu belajar mingguan yang rendah dengan minim dukungan tambahan.

Sementara itu, pola pembelajaran Intensive mencakup 29.2% dari total siswa, yang berarti hampir sepertiga siswa memiliki waktu belajar yang tinggi dan mendapatkan dukungan seperti tutoring atau parental support yang kuat.

Terakhir, pola pembelajaran Balanced memiliki proporsi terkecil, yaitu 10.4%, menunjukkan bahwa hanya sedikit siswa yang memiliki waktu belajar sedang dan aktif dalam kegiatan ekstrakurikuler.

Dari data ini, dapat disimpulkan bahwa sebagian besar siswa cenderung memilih pola pembelajaran yang minim intensitas dan dukungan tambahan, sehingga perlu adanya strategi untuk meningkatkan efektivitas pembelajaran bagi kelompok ini.

## **Histogram**

```python
plt.figure(figsize=(10, 6))
sns.histplot(df["GPA"], bins=30, kde=True, color="blue")

# Tambahkan judul dan label
plt.title("Distribusi Nilai GPA Siswa", fontsize=14)
plt.xlabel("GPA", fontsize=12)
plt.ylabel("Frekuensi", fontsize=12)

# Tampilkan plot
plt.show()
```

Histogram ini menunjukkan distribusi nilai GPA (Grade Point Average) siswa dalam dataset. Dari visualisasi ini, terlihat bahwa nilai GPA siswa tersebar dalam rentang 0 hingga 4.0, dengan mayoritas siswa memiliki nilai di kisaran 1.5 hingga 2.5.

Sebagian besar siswa memiliki GPA dalam rentang sedang, yang menunjukkan bahwa rata-rata performa akademik mereka tidak terlalu tinggi maupun terlalu rendah. Namun, ada juga sebagian kecil siswa dengan GPA yang sangat rendah (mendekati 0) dan sangat tinggi (mendekati 4.0).

## **Scatter Plot**

```python
# Membuat scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['StudyTimeWeekly'], df['GPA'], c='blue', alpha=0.5)

# Menambahkan garis regresi menggunakan seaborn
sns.regplot(x=df['StudyTimeWeekly'], y=df['GPA'], scatter=False, color='red')

# Menambahkan judul dan label sumbu
plt.title('Korelasi Waktu Belajar Mingguan vs GPA')
plt.xlabel('Waktu Belajar Mingguan (jam)')
plt.ylabel('GPA')

# Menampilkan plot
plt.show()
```

Scatter plot ini menggambarkan hubungan antara Waktu Belajar Mingguan (StudyTimeWeekly) dengan GPA siswa dalam bentuk sebaran titik data. Titik-titik biru pada grafik menunjukkan distribusi siswa berdasarkan waktu belajar mingguan dan GPA mereka. Sebagian besar siswa memiliki GPA antara 1.5 - 2.5, dengan waktu belajar yang bervariasi dari sangat sedikit hingga lebih dari 15 jam per minggu.

Garis merah menunjukkan tren regresi linier, yang memiliki kemiringan positif tetapi sangat kecil. Ini mengindikasikan bahwa ada hubungan positif yang lemah antara waktu belajar dan GPA—artinya, semakin lama siswa belajar, GPA cenderung meningkat, tetapi tidak secara signifikan.

## **Heatmap**

```python
# Heatmap (Seaborn): Korelasi Antar Variabel Numerik
plt.figure(figsize=(8, 5))
sns.heatmap(df[['StudyTimeWeekly', 'GPA']].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Heatmap Korelasi: Waktu Belajar vs GPA')
plt.show()
```

Heatmap di atas menunjukkan korelasi antara StudyTimeWeekly (Waktu Belajar Mingguan) dan GPA (Indeks Prestasi Kumulatif). Nilai korelasi sebesar 0.18 menunjukkan adanya hubungan positif yang lemah antara waktu belajar mingguan dan GPA. Artinya, semakin banyak waktu yang dihabiskan untuk belajar, kemungkinan GPA meningkat, tetapi pengaruhnya tidak terlalu kuat.

**Penyesuaian Metode Pembelajaran Berdasarkan Pola Belajar**

Untuk siswa dengan pola Minimalist, materi dapat disajikan dalam bentuk yang lebih singkat dan langsung ke inti pembelajaran, seperti melalui materi ringkas atau konten berbasis modul mandiri. Institusi pendidikan dapat mempertimbangkan kurikulum yang lebih fleksibel, yang memungkinkan siswa memilih metode belajar yang paling sesuai dengan kebutuhan mereka. **Pemberian Dukungan Pembelajaran yang Lebih Personal**

Pendekatan pembelajaran dapat disesuaikan dengan karakteristik masing-masing siswa. Dan siswa dengan pola Intensive bisa diberikan tantangan akademik yang lebih tinggi, seperti advanced coursework atau program akselerasi.
